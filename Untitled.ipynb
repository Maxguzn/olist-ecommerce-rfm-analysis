```python
import pandas as pd
orders = pd.read_csv('data/olist_orders_dataset.csv')
display(orders.head())
orders.info()
```

```python
date_columns = [
    'order_purchase_timestamp', 
    'order_approved_at', 
    'order_delivered_carrier_date', 
    'order_delivered_customer_date', 
    'order_estimated_delivery_date'
]
for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])

orders.info()
```

```python
print(orders.isnull().sum())
```

```python
orders = orders[orders['order_status'] == 'delivered']
orders.dropna(inplace=True)
print(orders.isnull().sum())
print(f'\nОсталось строк: {len(orders)}')
```

```python
customers = pd.read_csv('data/olist_customers_dataset.csv')
display(customers.head())
customers.info()
```

```python
orders_customers = pd.merge(
    orders,             
    customers,          
    on='customer_id',   
    how='inner'         
)
display(orders_customers.info())
print(f"Строк: {len(orders_customers)}")
print(f"Колонок: {len(orders_customers.columns)}")
```

```python
items = pd.read_csv('data/olist_order_items_dataset.csv')
print("Сырая таблица товаров:")
display(items.head())
full_table = pd.merge(
    orders_customers, 
    items, 
    on='order_id', 
    how='inner'
)
print(f"\nСтрок ДО объединения: {len(orders_customers)}")
print(f"Строк ПОСЛЕ объединения: {len(full_table)}")
print(f"Колонок теперь: {len(full_table.columns)}")
```

```python
payments = pd.read_csv('data/olist_order_payments_dataset.csv')
print("Сырая таблица оплат:")
display(payments.head())
final_table = pd.merge(
    full_table, 
    payments, 
    on='order_id', 
    how='inner'
)
print(f"\nСтрок ДО объединения с платежами: {len(full_table)}")
print(f"Строк ПОСЛЕ объединения: {len(final_table)}")
print(f"Колонок теперь: {len(final_table.columns)}")
```

```python
import sqlite3
conn = sqlite3.connect('olist_database.db')
final_table.to_sql(
    'ecommerce_data',   
    conn,                
    if_exists='replace', 
    index=False          
)

print("Данные успешно загружены в SQL-базу: olist_database.db")
```

```python
# Пишем многострочный SQL-запрос
rfm_query = """
SELECT 
    customer_unique_id, MAX(order_purchase_timestamp) AS last_purchase_date, COUNT(DISTINCT order_id) AS frequency, SUM(payment_value) AS monetary
FROM ecommerce_data
GROUP BY customer_unique_id
ORDER BY monetary DESC;
"""
rfm_data = pd.read_sql(rfm_query, conn)

print("Топ 10 наших лучших клиентов:")
display(rfm_data)
```

```python
import datetime

rfm_data['last_purchase_date'] = pd.to_datetime(rfm_data['last_purchase_date'])
snapshot_date = rfm_data['last_purchase_date'].max() + datetime.timedelta(days=1)

print(f"Дата отсчета для нашего анализа (Snapshot Date): {snapshot_date}\n")

rfm_data['recency'] = (snapshot_date - rfm_data['last_purchase_date']).dt.days

print("Готовая RFM-таблица (первые 5 клиентов):")
display(rfm_data[['customer_unique_id', 'recency', 'frequency', 'monetary']].head())
```

```python
r_labels = [5, 4, 3, 2, 1] 
rfm_data['R_Score'] = pd.qcut(rfm_data['recency'], q=5, labels=r_labels)

m_labels = [1, 2, 3, 4, 5]
rfm_data['M_Score'] = pd.qcut(rfm_data['monetary'], q=5, labels=m_labels)

def get_f_score(x):
    if x == 1:
        return 1
    elif x == 2:
        return 2
    elif x == 3:
        return 3
    elif x == 4:
        return 4
    else:
        return 5

rfm_data['F_Score'] = rfm_data['frequency'].apply(get_f_score)
rfm_data['RFM_Segment'] = rfm_data['R_Score'].astype(str) + rfm_data['F_Score'].astype(str) + rfm_data['M_Score'].astype(str)
rfm_data['RFM_Total_Score'] = rfm_data['R_Score'].astype(int) + rfm_data['F_Score'].astype(int) + rfm_data['M_Score'].astype(int)

print("Финальная RFM-таблица:")
display(rfm_data.head(10))
```

```python
rfm_data.to_csv('data/rfm_segments_final.csv', index=False)
print("Файл rfm_segments_final.csv успешно создан и сохранен в папке data.")
```
